## 6. Peer review and debrief · 5 minutes

Swap recommendations with your partner. Ask: **“What evidence did I overlook?
What could change my recommendation?”** Revise one sentence if needed.

| Question | Concept | In this case |
|---|---|---|
| Who is missing? | Selection bias | The filter removes nonpurchasing respondents. Even all responses leave out nonrespondents. |
| Could the mix explain this? | Simpson's paradox | Conversion rises within both store types but falls overall as more visitors go to busy stores. |
| Am I only looking for support? | Confirmation bias | Someone invested in the promotion could focus on happy purchasers or improved subgroup rates and overlook other evidence. |

**Simpson's paradox does not prove success.** It explains why an aggregate comparison
can differ from every within-group comparison. Here, purchases really fell from
84 to 48; the financial and causal questions remain unanswered.
