## Debrief · Two questions to remember · 5 minutes

Read this after discussing your recommendation.

| Habit | Name | Where it appeared |
|---|---|---|
| **Who is missing?** | Selection bias | The code excludes refunded respondents. Including them changes satisfaction from 90% to 77.1%, but nonrespondents and nonbuyers are still unrepresented. |
| **Could the mix explain this?** | Simpson's paradox | Overall conversion fell even though it increased for both new and returning visitors; the group sizes changed substantially. |

Selection bias concerns who gets represented in the data. Simpson's paradox is
an aggregation pattern, not a psychological bias or a calculation error.

**An accurate number can support an inaccurate conclusion.** The overall and subgroup
conversion rates are all correct. They answer different descriptive questions; neither
alone tells us what the promotion caused. Missing survey responses cannot be repaired
by silently filling in invented opinions.

**Connect to pytest:** an assertion can check that overall conversion is 40%, or that
37 of the 48 survey respondents gave a rating of 4 or 5. It cannot, by itself, prove that surveyed
people represent all visitors or that the promotion caused a change. A passing test is
only as informative as the claim it actually checks.
